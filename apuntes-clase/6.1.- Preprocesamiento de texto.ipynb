# 6.1.- Preprocesamiento de texto

El primer paso en cualquier proyecto de **Procesamiento del Lenguaje Natural (NLP)** es **limpiar y normalizar el texto**.  
El lenguaje humano es ruidoso: contiene tildes, mayúsculas, signos, URLs y otras inconsistencias que los modelos no entienden directamente.

En este notebook aprenderás a:
1. Limpiar texto usando expresiones regulares (`re`)
2. Tokenizar palabras con `nltk`
3. Eliminar *stopwords* irrelevantes
4. (Opcional) Lematizar palabras con `spaCy`

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Descargar recursos necesarios
nltk.download('punkt_tab')
nltk.download("stopwords")
```

## Limpieza del texto (Text Cleaning)

Antes de tokenizar, debemos eliminar todo lo que **no aporte significado**:
- Convertir a minúsculas  
- Quitar signos de puntuación y caracteres especiales  
- Eliminar tildes o acentos  
- Quitar espacios repetidos  

Usaremos expresiones regulares (`re.sub`) para normalizar el texto.

```python
# Ejemplo de texto crudo
texto_original = "¡Hola, Mundo!!! Qué día más BONITO... ☀️🌍"

# 1. Convertir a minúsculas
texto = texto_original.lower()

# 2. Eliminar signos y caracteres especiales (excepto letras y espacios)
texto = re.sub(r"[^a-záéíóúüñ ]", "", texto)

# 3. Quitar espacios múltiples
texto = re.sub(r"\s+", " ", texto).strip()

print("Texto original:", texto_original)
print("Texto limpio:", texto)
```

## Tokenización

La **tokenización** consiste en dividir el texto en **unidades pequeñas** llamadas *tokens*,  
normalmente palabras. Esto permite al modelo analizar y contar cada palabra de forma individual.

Usaremos el tokenizador de `nltk`, aunque existen opciones más avanzadas como `spaCy`.

```python
# Tokenizar texto limpio
tokens = word_tokenize(texto)

print("Tokens obtenidos:")
print(tokens)
print("\nNúmero total de tokens:", len(tokens))
```

## Eliminación de *Stopwords*

Las *stopwords* son palabras muy frecuentes (como “el”, “de”, “y”, “en”)  
que **no aportan significado relevante** en tareas como clasificación o análisis de sentimiento.

Ejemplo:  
"El gato duerme en la casa" → ["gato", "duerme", "casa"]

```python
# Cargar lista de stopwords en español
stop_es = set(stopwords.words("spanish"))

# Filtrar tokens
tokens_filtrados = [t for t in tokens if t not in stop_es]

print("Tokens sin stopwords:")
print(tokens_filtrados)
print("\nNúmero de tokens finales:", len(tokens_filtrados))
```

## (Opcional) Lematización

La **lematización** reduce cada palabra a su forma base (*lema*).  
Por ejemplo:
- "corriendo" → "correr"
- "mejores" → "bueno"

Esto ayuda a **unificar** palabras con el mismo significado gramatical.

```python
import spacy
nlp = spacy.load("es_core_news_sm")

doc = nlp("Los gatos corrían rápidamente por las calles.")
lemmas = [token.lemma_ for token in doc]
print("Lemas:", lemmas)
```
